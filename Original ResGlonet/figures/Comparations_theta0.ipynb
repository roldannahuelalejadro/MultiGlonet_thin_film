```python
import pandas as pd
import ast
import matplotlib.pyplot as plt
import numpy as np
```

```python
valores_capas_1 = ref_index[0]
espesores_1 = thicknesses[0]

# Definir los valores y espesores de las capas para la segunda estructura
valores_capas_2 = ref_index[50]
espesores_2 = thicknesses[50]

# Definir los valores y espesores de las capas para la tercera estructura
valores_capas_3 = ref_index[100]
espesores_3 = thicknesses[100]

# Crear un colormap
colormap = plt.get_cmap('turbo')

# Definir el rango de valores
valor_min = 1
valor_max = 2.6

# Asignar colores a cada valor basado en el colormap
colores_1 = [colormap((valor - valor_min) / (valor_max - valor_min)) for valor in valores_capas_1]
colores_2 = [colormap((valor - valor_min) / (valor_max - valor_min)) for valor in valores_capas_2]
colores_3 = [colormap((valor - valor_min) / (valor_max - valor_min)) for valor in valores_capas_3]

# Crear la figura y el eje para todas las estructuras, con una relación de aspecto más ancha que alta
fig, (ax2, ax1) = plt.subplots(2, 1, figsize=(10, 6), gridspec_kw={'height_ratios': [2, 1]})

# Gráfico de error cuadrático mínimo en ax1
ax1.plot(Flancos, Error_0, "o-", alpha=0.8, label=r'Error en $\theta$ = 0')
ax1.plot(Flancos[0], Error_0[0], "o", color="Red", label="Error iteración 0")
ax1.plot(Flancos[50], Error_0[50], "o", color="Red", label="Error iteración 50")
ax1.plot(Flancos[100], Error_0[100], "o", color="Red", label="Error iteración 100")
ax1.set_xlabel(r"Posición del flanco ($nm$)")
ax1.set_ylabel('Error cuadrático mínimo')
# ax1.legend()
# ax1.set_title('Error Cuadrático Mínimo')  # Tamaño 12x3 para que sea más ancho que alto

# Calcular la suma acumulada de los espesores para todas las estructuras
espesores_acumulados_1 = np.cumsum(espesores_1)
espesores_acumulados_2 = np.cumsum(espesores_2)
espesores_acumulados_3 = np.cumsum(espesores_3)

# Definir el ancho de las barras y la separación entre las estructuras
ancho_barra = 0.2
separacion = 0.3

# Graficar las barras para la primera estructura
for i in range(len(espesores_1)):
    ax2.bar(1 - ancho_barra - separacion, espesores_1[i], width=ancho_barra,
           bottom=espesores_acumulados_1[i] - espesores_1[i],
           color=colores_1[i], edgecolor='black', linewidth=0)

# Graficar las barras para la segunda estructura
for i in range(len(espesores_2)):
    ax2.bar(1, espesores_2[i], width=ancho_barra,
           bottom=espesores_acumulados_2[i] - espesores_2[i],
           color=colores_2[i], edgecolor='black', linewidth=0)

# Graficar las barras para la tercera estructura
for i in range(len(espesores_3)):
    ax2.bar(1 + ancho_barra + separacion, espesores_3[i], width=ancho_barra,
           bottom=espesores_acumulados_3[i] - espesores_3[i],
           color=colores_3[i], edgecolor='black', linewidth=0)

# Configurar las etiquetas y el título
ax2.set_ylabel(r'Espesor (nm)')
ax2.set_xticks([1 - ancho_barra - separacion, 1, 1 + ancho_barra + separacion])
ax2.set_xticklabels(['', '', ''])  # Nombres de las estructuras
# ax2.set_title('Comparación de Estructuras Fotónicas')

# Mostrar la barra de colores que representa el mapa de valores
sm = plt.cm.ScalarMappable(cmap=colormap, norm=plt.Normalize(valor_min, valor_max))
sm.set_array([])
fig.colorbar(sm, ax=ax2, label='Índice de refracción')

# Ajustar el diseño para evitar cortes en los márgenes
plt.tight_layout()

reflection_1 = reflections[0]
reflection_2 = reflections[50]
reflection_3 = reflections[100]

# Generar los valores de lambdas_k
lambdas_k = np.linspace(400, 700, 400)

# Definir los targets para cada conjunto
targets_1 = np.ones(400)
targets_2 = np.ones(400)
targets_3 = np.ones(400)

# Modificar los targets según los índices dados
targets_1[:0 * 4] = 0
targets_2[:50 * 4] = 0
targets_3[:100 * 4] = 0

# Crear la figura con tres subplots en formato de columna
fig, axs = plt.subplots(1, 3, figsize=(10, 3), sharex=True)

# Colores para cada gráfico
colors = ['blue', 'orange', 'green']

# Graficar las reflexiones en cada subplot con el mismo ylim
for ax, reflection, targets, color, title in zip(axs, 
                                                 [reflection_1, reflection_2, reflection_3], 
                                                 [targets_1, targets_2, targets_3], 
                                                 colors, 
                                                 ['Reflexión 1 y Objetivo', 'Reflexión 2 y Objetivo', 'Reflexión 3 y Objetivo']):
    ax.plot(lambdas_k, targets, ".-", color='red',alpha = 0.5,  label='Objetivo')  # Formato "o-"
    ax.plot(lambdas_k, reflection, label='Simulado', color=color)
    ax.set_ylim(-0.1, 1.1)
    ax.set_ylabel('Reflectancia')
    ax.legend()

# Configurar el eje x compartido
axs[0].set_xlabel('Longitud de onda (nm)')
axs[1].set_xlabel('Longitud de onda (nm)')
axs[2].set_xlabel('Longitud de onda (nm)')

# Ajustar el diseño para evitar solapamientos
plt.tight_layout()

# Mostrar la gráfica
plt.show()
```
